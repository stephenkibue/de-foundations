### Q1 & Q2: How many rows and columns are in the clean dataset, what are their data types, and are there any missing values?

```python
# Load with date parsing enabled to handle the date column correctly
df = pd.read_csv(clean_path, parse_dates=["date"])

# Q1: Rows and columns
print("Dimensions (Rows, Columns):", df.shape)

# Q2: Data types and non-null counts
df.info()
```

### How many total eggs were produced across all records in the dataset?

```python
# Calculate total eggs
total_eggs = df["eggs"].sum()
print(f"Total eggs: {total_eggs:,}")
```

### How many clean production records exist for each flock?

```python
# Count records per flock
df["flock"].value_counts()
```

### On which days did Flock C lay fewer than 410 eggs?

```python
# Filter for Flock C with less than 410 eggs
flock_c_low = df[(df["flock"] == "C") & (df["eggs"] < 410)]
flock_c_low[["date", "flock", "eggs"]]
```

### On how many days did Flock A lay more than 1,000 eggs, and which days were they?

```python
# Filter for Flock A with more than 1000 eggs
flock_a_high = df[(df["flock"] == "A") & (df["eggs"] > 1000)]
print(f"Number of days: {len(flock_a_high)}")
flock_a_high[["date", "flock", "eggs"]]
```

### How many records did Otieno enter from 10 October onwards, and how many eggs do they add up to in total?

```python
# Filter for Otieno's records on or after October 10
otieno_oct10 = df[(df["recorded_by"] == "Otieno") & (df["date"] >= "2026-10-10")]

print("Record count:", len(otieno_oct10))
print("Total eggs:", otieno_oct10["eggs"].sum())
otieno_oct10
```

### How many eggs did Flock A lay in the week of 8 to 14 October 2026?

```python
# Filter Flock A for dates between October 8 and October 14 (inclusive)
flock_a_week = df[
    (df["flock"] == "A") & 
    (df["date"].between("2026-10-08", "2026-10-14"))
]

print("Record count:", len(flock_a_week))
print("Total eggs:", flock_a_week["eggs"].sum())
flock_a_week[["date", "flock", "eggs"]]
```
